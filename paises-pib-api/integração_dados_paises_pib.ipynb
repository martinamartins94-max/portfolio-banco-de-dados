Importação das bibliotecas necessárias

```python
import json
import requests
import pandas as pd
from dataclasses import dataclass
```

Coleta de dados da API REST

```python
r = requests.get("http://127.0.0.1:5000/country")
data = r.json()

linhas = []
for codigo, info in data.items():
    linhas.append({
        "codigo" : codigo,
        "nome" : info["name"],
        "codigo_moeda" : info["currency"],
        "populacao" : info ["population"],
        "capital" : info["capital"],
        "area" : info["area"]
    })

    df = pd.DataFrame(linhas)
    print(df.head())

    df.to_csv("paises_api.csv", index=False)
```

    Leitura dos arquivos CSV paises_api (obtido por meio da API REST) e gdp_data com dados complementares sobre o PIB dos países entre 1960 a 2022

```python
paises = pd.read_csv('paises_api.csv')
gdp = pd.read_csv('gdp_data.csv')
```

Seleção da coluna PIB

```python
colunas_gdp = ['Country Code'] + [str(ano) for ano in range(1960, 2022)]
gdp = gdp[colunas_gdp]
```

Inserir o mesmo índice nas duas tabelas (código do país)

```python
paises = paises.set_index('codigo')          # coluna do paises_api
gdp = gdp.set_index('Country Code')         # coluna do gdp_data
```

Concatenação dos dados em um único dataframe

```python
df_final = pd.concat([paises, gdp], axis=1)
```

Salvar em CSV

```python
df_final.to_csv('paises_api_com_pib_concat.csv', index=False)
```
